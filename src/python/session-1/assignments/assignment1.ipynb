Input Validation

```python
email="Amit_ml@gmail.edu"
if email.find("@") == 0 and email.find(".") > email.find("@"):
    print("Valid email ")
else:
    print("Invalid email ")
```

Extract Username

```python
email="Amit_ml@gmail.edu"
At=email.find("@")
print(email[0:At])
```

Extract Domain

```python
email="Amit_ml@gmail.edu"
At=email.find("@")
Dot=email.find(".")
print(email[At+1:Dot])
```

Check for Domain Ending

```python
email="Amit_ml@gmail.edu"
if(email[-4:]==".com"): print("Commercial Domain ")
elif(email[-4:]==".edu"): print("Educational Domain ")
else:print("Other Domain")
```
