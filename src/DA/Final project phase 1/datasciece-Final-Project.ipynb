```python
# trying to fill missing values in my dataset using for loop

df_full = df.copy()
filled_count = 0

print("\ndisplay missing values and their data types")
for col in df_full.columns:
    if df_full[col].isnull().any():
        print(f"{col}: {df_full[col].dtype} (Missing: {df_full[col].isnull().sum()})")


for column in df_full.columns:
    if df_full[column].isnull().any():
        
        # numeric columns (float64, int64): fill with mean  
        if df_full[column].dtype in ['float64', 'int64']:
            mean_value = df_full[column].mean()
            missing_before = df_full[column].isnull().sum()
            df_full[column] = df_full[column].fillna(mean_value)
            filled_count += missing_before
            print(f"✅ column '{column}': {missing_before} Missing value filled using mean ({mean_value:.2f})")
        
        # categorical columns (object, string, category): fill with mode
        elif df_full[column].dtype in ['object', 'string', 'category']:
            mode_value = df_full[column].mode()[0]
            missing_before = df_full[column].isnull().sum()
            df_full[column] = df_full[column].fillna(mode_value)
            filled_count += missing_before
            print(f"✅ column '{column}': {missing_before} Missing value filled using mode ('{mode_value}')")
        
        # numeric columns (float64, int64): fill with mean
        else:
            try:
                mean_value = df_full[column].mean()
                missing_before = df_full[column].isnull().sum()
                df_full[column] = df_full[column].fillna(mean_value)
                filled_count += missing_before
                print(f"✅ column '{column}': {missing_before} Missing value filled using mode ('{mode_value}')")
            except:
                print(f"column '{column}': لا يمكن تعبئة القيم المفقودة (نوع البيانات: {df_full[column].dtype})")



```

```python
## create a function to know the average users ages
def get_age_statics(df, age_column='member_age'):
    """
    this function calculate and display comprehensive age statistics for users
    
    Parameters
    df : pandas DataFrame
        The DataFrame containing the age column
    age_column : str
        The name of the age column 'member_age'
    
    Returns:
    dict : Dictionary containing all age statistics
    """
    
    print("Age Report")
    stats = {
        'count': len(df),
        'mean': df[age_column].mean(),
        'median': df[age_column].median(),
        'min': df[age_column].min(),
        'max': df[age_column].max(),
        'std': df[age_column].std(),
        'q1': df[age_column].quantile(0.25),
        'q3': df[age_column].quantile(0.75),
        'skewness': df[age_column].skew(),
        'kurtosis': df[age_column].kurtosis()
    }
    
    print(f"\n Basic :")
    print(f"  Total users: {stats['count']:,}")
    print(f" Average age: {stats['mean']:.2f} years")
    print(f"  Median age: {stats['median']:.2f} years")
    print(f"  Age range: {stats['min']} to {stats['max']} years")
    print(f"  Standard deviation: {stats['std']:.2f}")
    print(f"  25th percentile: {stats['q1']:.2f}")
    print(f"   75th percentile: {stats['q3']:.2f}")
    print(f" Skewness: {stats['skewness']:.2f}")
    print(f"   Kurtosis: {stats['kurtosis']:.2f}")
    
    print("\nAge Groups")
    bins = [20, 30, 40, 50, 60, 70, 80, 90, 100]
    labels = ['20-29', '30-39', '40-49', '50-59', '60-69', '70-79', '80-89', '90+']
    df['age_group'] = pd.cut(df[age_column], bins=bins, labels=labels, right=False)
    print("\nAge Group Distribution:")
    age_distribution = df['age_group'].value_counts().sort_index()
    age_percentage = (age_distribution / len(df)) * 100
    for group in labels:
        if group in age_distribution.index:
            count = age_distribution[group]
            pct = age_percentage[group]
            print(f"   - {group}: {count:,} users ({pct:.1f}%)")
    
    most_common_age = df[age_column].mode()[0]
    most_common_count = df[df[age_column] == most_common_age].shape[0]
    print(f"\nMost common age: {most_common_age} years ({most_common_count:,} users)")
    
    # Check outliers ages > 100
    outliers = df[df[age_column] > 100]
    if len(outliers) > 0:
        print(f"\n ages > 100: {len(outliers)} users")
    
    return stats

```
