```python
def arra(arr , row : int = -1 ,col : int = -1 ):
    try:
        arr= np.array(arr)
        return arr.reshape(row , col)
    except (TypeError , ValueError) as e:
        raise ValueError (f"Invalid reshape :{e}")

lis = [1,2,3,4,5]
print(arra(lis))
```
