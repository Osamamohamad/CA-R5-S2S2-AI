```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([2.6, 4.7, 5.1, 4.2, 5.3])
```

```python
n = len(x)

w=0
b=0
np.random.seed(42)
w = np.random.rand()
b = np.random.rand()
learning_rate = 0.01
epochs = 10000
costs = []
error = 0
print("Initial weight:", w)
print("Initial bias:", b)
print("Learning rate:", learning_rate)
print("Epochs:", epochs)
```

```python
for epoch in range(epochs):
    y_pred = w * x + b
    error = y_pred - y

    dw = np.mean(error * x)
    db = np.mean(error)
    
    w -= learning_rate * dw
    b -= learning_rate * db

    cost = 0.5 * np.mean(error ** 2)
    costs.append(cost)

    if epoch % 10 == 0 :
        print(f"Epoch {epoch}:  cost [{cost:.6f}] Weight {w:.6f}, Bias {b:.6f}")
```

```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

deta = pd.read_csv("Salary_Data.csv")
```

```python
deta.head(10)
```

```python
plt.figure(figsize=(10, 5))
sns.pairplot(data = deta , x_vars=["YearsExperience"], y_vars=["Salary"], size = 5, kind='reg')
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Scatter Plot of Years of Experience vs Salary")
```

```python
x=deta.iloc[:, :-1]
y=deta.iloc[:, 1]
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
linear_model = LinearRegression()
linear_model.fit(x_train, y_train)

y_pred = linear_model.predict(x_train)
```

```python
plt.scatter(x_train, y_train, color='blue', label='Training Data')
plt.plot(x_train, y_pred, color='red', label='Regression Line')
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Scatter Plot of Years of Experience vs Salary")
plt.legend()
plt.show()
```

```python
y_pred_test = linear_model.predict(x_test)
plt.scatter(x_test, y_test, color='green', label='Test Data')
plt.plot(x_test, y_pred_test, color='orange', label='Test Regression Line')
plt.xlabel("Years of Experience")
plt.ylabel("Salary")
plt.title("Scatter Plot of Years of Experience vs Salary")
plt.legend()
plt.show()
```
